```python
import duckdb

# 1. Path to your DuckDB file
db_path = "data/processed/latam_bank.duckdb"

# 2. Connect to the database
# Set read_only=True if you only need to query data (prevents file locking issues)
con = duckdb.connect(db_path, read_only=True)

try:
    # 3. Execute a query and fetch the results as a Pandas DataFrame
    print("Fetching top 5 contact reasons...")
    df = con.execute("""
        SELECT contact_reason, COUNT(*) as count 
        FROM call_center_interactions 
        GROUP BY contact_reason 
        ORDER BY count DESC 
        LIMIT 5
    """).df()
    
    print(df)

finally:
    # 4. Always close the connection when done
    con.close()
```
